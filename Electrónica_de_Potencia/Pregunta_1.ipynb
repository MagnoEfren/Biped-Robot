La corriente de carga (¿La corriente de carga es continua?)

La corriente promedio y rms en la carga.

La corriente promedio y rms en cada tiristor.

La corriente promedio y rms de la fuente.

El valor alpha crítico para la corriente de carga (valor cuando la corriente de carga es 0).

```python
import numpy as np

alpha = 30 #180/6 # Ángulo de disparo
Vs = 240 #
R = 2.5 # ohms
L = 6.5e-3# Inductancia
f = 60 # Frecuencia en Hz
E = 10 # FUENTE
Vm = Vs*np.sqrt(2)
alpha = np.radians(alpha)
omega = 2*np.pi*f

Icd = ((2*Vm*np.cos(alpha))/np.pi - E)/R

a2 = (2*Vm/np.pi)*(np.cos(3*alpha)/3 - np.cos(1*alpha)/1)
b2 = (2*Vm/np.pi)*(np.sin(3*alpha)/3 - np.sin(1*alpha)/1)
V2 = np.sqrt(a2**2 + b2**2)
a4 = (2*Vm/np.pi)*(np.cos(5*alpha)/5 - np.cos(3*alpha)/3)
b4 = (2*Vm/np.pi)*(np.sin(5*alpha)/5 - np.sin(3*alpha)/3)
V4 = np.sqrt(a4**2 + b4**2)
Z2 = np.sqrt(R**2 + (2*omega*L)**2)
Z4 = np.sqrt(R**2 + (4*omega*L)**2)
I2 = V2/Z2
I4 = V4/Z4
I2_rms = I2/np.sqrt(2)
I4_rms = I4/np.sqrt(2)
Irms = np.sqrt(Icd**2 + I2_rms**2 + I4_rms**2)

# Verificar si es continua o no:
Z = np.sqrt(R**2 + (omega*L)**2)
theta = np.arctan(omega*L/R)

e_x = np.exp(-(R/L)*(np.pi/omega))
I_Lo = np.sqrt(2)*Vs/Z*((-np.sin(alpha-theta)-np.sin(alpha-theta)*e_x)/(1-e_x))- E/R


## Corriente media y rms en los SCR
Icd_SCR = Icd/2
Imrs_SCR = Irms/np.sqrt(2)

####
Irms_fuente = Irms
Idc_fuente = 0
# Encontramos el alpha critico:
   #  trabajamos de 0 a 180

list_angulos_posi = []
list_angulos_nega = []
for i in range(0,180+1):
    alpha_critico = np.radians(i)
    e_x1 = np.exp(-(R/L)*(np.pi/omega))
    I_Lo1 = np.sqrt(2)*Vs/Z*((-np.sin(alpha_critico-theta)-np.sin(alpha_critico-theta)*e_x1)/(1-e_x1))- E/R
    if I_Lo1 >0:
        list_angulos_posi.append(i)
    else:
        list_angulos_nega.append(i)

print('La corriente de carga', round(I_Lo, 5) ,'A', 'Es mayor que 0 --> ES CONTINUA')

print('Corriente rms  de la carga: ', round((Irms), 5), 'A')
print('Corriente promedio  de la carga: ', round((Icd), 5), 'A')


print('Valor corriente media en los SCR:  ', round((Icd_SCR), 5), 'A')
print('Valor corriente rms en los SCR:  ', round((Imrs_SCR), 5), 'A')


print('Valor corriente promedio en la fuente:  ', round((Idc_fuente), 5), 'A')
print('Valor corriente rms de la fuente:  ', round((Irms_fuente), 5), 'A')

print('El valor alpha crítico: ')
print('Angulos donde es continua: ',list_angulos_posi)
print('Angulos donde no es continua: ',list_angulos_nega)
```
