Un R.M.O.C. controlado presenta una carga RL con R=5.5 ohms, L=36 mH. La tensión eficaz del generador es de 240V a 50Hz. Si alpha=pi/3, determine lo siguiente:

Corriente estable de carga (¿La corriente de carga es continua?)

Corriente promedio y rms en la carga.

Potencia absorbida.

Factor de potencia.


```python
import numpy as np
alpha = 60 #180/3 # Ángulo de disparo
Vs = 240 #
R = 5.5 # ohms
L = 36e-3# Inductancia
f = 50 # Frecuencia en Hz
E = 0 # FUENTE
Vm = Vs*np.sqrt(2)
alpha = np.radians(alpha)
omega = 2*np.pi*f

Icd = ((2*Vm*np.cos(alpha))/np.pi - E)/R

a2 = (2*Vm/np.pi)*(np.cos(3*alpha)/3 - np.cos(1*alpha)/1)
b2 = (2*Vm/np.pi)*(np.sin(3*alpha)/3 - np.sin(1*alpha)/1)
V2 = np.sqrt(a2**2 + b2**2)
a4 = (2*Vm/np.pi)*(np.cos(5*alpha)/5 - np.cos(3*alpha)/3)
b4 = (2*Vm/np.pi)*(np.sin(5*alpha)/5 - np.sin(3*alpha)/3)
V4 = np.sqrt(a4**2 + b4**2)
Z2 = np.sqrt(R**2 + (2*omega*L)**2)
Z4 = np.sqrt(R**2 + (4*omega*L)**2)
I2 = V2/Z2
I4 = V4/Z4
I2_rms = I2/np.sqrt(2)
I4_rms = I4/np.sqrt(2)
Irms = np.sqrt(Icd**2 + I2_rms**2 + I4_rms**2)

# Verificar si es continua o no:
Z = np.sqrt(R**2 + (omega*L)**2)
theta = np.arctan(omega*L/R)


e_x = np.exp(-(R/L)*(np.pi/omega))
I_Lo = np.sqrt(2)*Vs/Z*((-np.sin(alpha-theta)-np.sin(alpha-theta)*e_x)/(1-e_x))- E/R

#Ptencia absorbida
Pabs = (Irms**2)*R

# Factor de potencia
S = Vs*Irms # Is=Irms

FP = Pabs/S

print('La corriente de carga', round(I_Lo, 5) ,'A', '- Es mayor que 0 --> ES CONTINUA')
print('Corriente promedio  de la carga: ', round((Icd), 5), 'A')
print('Corriente rms  de la carga: ', round((Irms), 5), 'A')
print('Potencia absorbida: ', round(Pabs,5), 'W')
print('Factor de Potencia: ', round(FP,5))

```
