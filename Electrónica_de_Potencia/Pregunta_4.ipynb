Un controlador monofásico tiene una fuente de Vs=220V, f=60Hz; y una carga RL de R=15 ohms y L=46 mH. Si se tiene un ángulo de disparo de pi/3, determinar lo siguiente:

El valor del ángulo de extinción: beta.

La corriente promedio y rms de carga.

La corriente promedio y rms en cada SCR (caso el circuito sea diseñado con 2 SCRs)

La potencia absorbida.

El factor de potencia.


```python


# Circuito convertidor AC/AC  con carga  RL
from numpy import radians, pi, cos, sin, sqrt
import scipy.integrate as integrate
import numpy as np
###### DATOS ##########
Vcd = 0 #valor medio del voltaje de salida será igual a 0
V_rms = 220 #Voltaje rms de entrada
Vs = V_rms # Voltaje entrada (Voltaje eficaz)
R = 15 # Resistencia  de carga en ohm (R)
alpha =  60  #180/3 # Ángulo en grados sexagesimales
L = 46e-3  # Inductancia  diferente de 0 (Carga RL)
f = 60  # Frecuencia en Hz
beta = 180

#######################
alpha = np.radians(alpha) # Convertimos el angulo a radianes
omega = 2*np.pi*f
Vm = Vs*np.sqrt(2)  # Voltaje pico en la entrada
theta = np.arctan(omega*L/R)
Z = np.sqrt((R**2)+((omega*L)**2))

# Expresión de la corriente para la primera mitad del periodo #exp=e**((R/(omega L))* (alpha-wt))
# io=(Vm/Z)*((sin(wt-theta))-exp (sin(alpha-theta)))
# La corriente eficaz de carga #calculo beta:
io=19 #valor al azar
while io>0:
    beta_rad = radians(beta)
    exp = np.exp((R/(omega*L))*(alpha-beta_rad))
    io = (Vm/Z)*((sin(beta_rad-theta))-exp* (sin(alpha-theta)))
    beta = beta + 0.01

beta = radians(beta)
#Corriente eficaz de carga
int_rms=integrate.quad(lambda x: ((Vm/Z)*((sin(x-theta)) - (np.exp((R/(omega*L))*(alpha-x)))*(sin(alpha-theta))))**2, alpha, beta)
Irms= np.sqrt(int_rms[0]/pi)



# La corriente eficaz en los SCR
Irms_SCR= Irms/np.sqrt(2)

# La corriente media en los SCR
Imed=integrate.quad(lambda x: ((Vm/Z)*((sin(x-theta)) - (np.exp((R/(omega*L))*(alpha-x)))*(sin(alpha-theta))))**1, alpha, beta)
Imed_SCR=(Imed[0]/(2*pi))

# La potencia entregada a la carga
Pabs = (Irms**2)*R

# El factor de potencia
S = Vs*Irms
FP = Pabs/S



print(f"ángulo de exitación de Beta: {round(np.rad2deg(beta), 5)}°")
print("La corriente eficaz de carga: ",round(Irms, 5), "A")
print('Corriente promedio  de la carga: ', 0, 'A')
print("La corriente eficaz en los SCR: ",round(Irms_SCR, 5), "A")
print("La corriente media en los SCR: ",round(Imed_SCR, 5), "A")
print("Potencia absorbida por la carga: ",round(Pabs, 5), "W")
print("El factor de potencia: ",round(FP,5))
```
